```python
from langgraph.graph import StateGraph, START, END
from langchain_mistralai import ChatMistralAI
from typing import TypedDict
from dotenv import load_dotenv
import os
```

```python
load_dotenv()  # Load environment variables from .env file
if not os.getenv("MISTRAL_API_KEY"):
    raise RuntimeError("MISTRAL_API_KEY is missing. Add it to the .env file in the lang folder.")
```

```python
model = ChatMistralAI(model="mistral-large-latest", temperature=0)
```

```python
class LLMState(TypedDict):
    user_input: str
    model_response: str
```

```python
def LLMqa(state: LLMState):
    ques = state['user_input']
    prompt = f"Answer the following question: {ques}"
    response = model.invoke(prompt)
    state['model_response'] = response.content
    return state
```

```python
graph = StateGraph(LLMState)
graph.add_node("getUserInput", LLMqa)
graph.add_edge(START, "getUserInput")
graph.add_edge("getUserInput", END)
workflow = graph.compile()
```

```python
initial_state = {"user_input": "What is the capital of France?"}
final_state = workflow.invoke(initial_state)
print(final_state["model_response"])
```
