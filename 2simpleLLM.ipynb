```python
from langgraph.graph import StateGraph, MessagesState, START, END
from langchain_google_genai import ChatGoogleGenerativeAI
from typing import TypedDict
from dotenv import load_dotenv
```

```python
load_dotenv()  
```

```python
model = ChatGoogleGenerativeAI(model="gemini-3.5-flash-lite")
```

```python
class LLMState(TypedDict):
    user_input: str
    model_response: str
```

```python
def LLMqa(state: LLMState):
    ques = state['user_input']
    prompt = f"Answer the following question: {ques}"
    response = model.invoke(prompt)
    state['model_response'] = response.content
    return state
```

```python
graph = StateGraph(LLMState)
graph.add_node("getUserInput", LLMqa)
graph.add_edge(START, "getUserInput")
graph.add_edge("getUserInput", END)
workflow = graph.compile()
```

```python
initial_state = {"user_input": "hello?"}
final_state = workflow.invoke(initial_state)
```
